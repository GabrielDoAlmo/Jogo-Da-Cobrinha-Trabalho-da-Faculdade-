```python
#Questao 5 de forma complexa
frase = input("digite uma frase\n").lower().split()
palavra_buscada = input("qual palavra voce gostaria de buscar?\n").lower()
contador = 0
for palavra in frase:
  if palavra == palavra_buscada:
   contador += 1
if contador >0:
  print(contador)
else:
  print("digite uma palavra que contem na frase")
```

```python
#QUESTAO6
frase = input("digite a sua frase\n").lower()
palavra_antiga = input(f"qual palavra deseja selecionar da frase para ser trocada?\n{frase}\n")
palavra_nova = input("\nqual palavra deseja colocar no local da anterior ?\n")
nova_frase = frase.replace (palavra_antiga,palavra_nova)
print(nova_frase)
```

```python
#QUESTAO9
lista_original = [
    "Esta é uma frase bem curtinha.",
    "Esta é uma frase muito mais longa que com certeza tem mais de dez palavras para testarmos o nosso código.",
    "Outra frase pequena.",
    "Vamos colocar mais uma frase gigante aqui só para garantir que o nosso contador de palavras vai funcionar direitinho na hora de ler."
]
lista_vazia = []
for quantidade in lista_original:
  contagem = len(quantidade.split())
  if contagem >10:
    lista_vazia.append(quantidade)
print(lista_vazia)
```
